```python
import importlib

packages = [
    "numpy", "pandas", "scipy", "sklearn", "statsmodels",
    "matplotlib", "seaborn", "plotly", "ipywidgets",
    "mlxtend", "sympy", "networkx", "PIL", "missingno",
    "openpyxl", "xlrd", "bs4", "html5lib", "sqlite3", "tabulate",
]

for name in packages:
    try:
        module = importlib.import_module(name)
        version = getattr(module, "__version__", getattr(module, "sqlite_version", ""))
        print(f"OK      {name:<12} {version}")
    except Exception as e:
        print(f"FAILED  {name:<12} {type(e).__name__}: {e}")
```

```python
# Can pandas write and read Excel, HTML and SQL? (tests each library pairing)
import io
import pandas as pd

sample = pd.DataFrame({"a": [1, 2], "b": [3, 4]})

def check(label, test):
    try:
        test()
        print(f"OK      {label}")
    except Exception as e:
        print(f"FAILED  {label}: {type(e).__name__}: {e}")

def excel():
    buffer = io.BytesIO()
    sample.to_excel(buffer, index=False)
    buffer.seek(0)
    assert pd.read_excel(buffer).shape == (2, 2)

def html():
    tables = pd.read_html(io.StringIO(sample.to_html(index=False)), flavor="bs4")
    assert tables[0].shape == (2, 2)

def sql():
    import sqlite3
    engine = sqlite3.connect(":memory:")
    sample.to_sql("data", engine, index=False)
    assert pd.read_sql("SELECT * FROM data", con=engine).shape == (2, 2)

check("Excel read/write", excel)
check("HTML table read", html)
check("SQL write/read (SQLite)", sql)
```
