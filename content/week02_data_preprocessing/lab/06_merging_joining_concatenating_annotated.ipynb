# Merging, Joining, and Concatenating

There are 3 main ways of combining DataFrames together: Merging, Joining and Concatenating. In this lecture we will discuss these 3 methods with examples.

____

## About this notebook

Data often arrives in several tables. This notebook shows three ways to combine them:
* **Concatenate** (`pd.concat`): stack tables on top of each other, or side by side.
* **Merge** (`pd.merge`): match rows that share the same value in a *key* column, like joining tables in SQL.
* **Join** (`.join`): combine two tables by matching their row labels (index).

**How to use it:** click a code cell and press **Shift + Enter** to run it. Run the cells from top to bottom, because later cells use things created in earlier ones. The short note above each code cell tells you what it does.

## Concatenation

Concatenation basically glues together DataFrames. Keep in mind that dimensions should match along the axis you are concatenating on. You can use **pd.concat** and pass in a list of DataFrames to concatenate together:

`pd.concat` glues the tables together. By default it stacks them **on top of each other**, giving one table of 12 rows.

```python
pd.concat([df1,df2,df3])
```

With `axis=1` the tables are placed **side by side** instead. pandas lines them up by row label, and because each table has different row labels, most of the result is **NaN**. This shows that concatenation works well only when the labels line up.

```python
pd.concat([df1,df2,df3],axis=1)
```

## Merging

The **merge** function allows you to merge DataFrames together using a similar logic as merging SQL Tables together. For example:

`pd.merge` combines the two tables by matching rows that have the same value in the `key` column, like a join in SQL. `how='inner'` keeps only keys found in **both** tables, which here is all of them.

```python
pd.merge(left,right,how='inner',on='key')
```

Or to show a more complicated example:

A harder example: now rows must match on **two** key columns, `key1` **and** `key2`, and not every combination appears in both tables.

```python
left = pd.DataFrame({'key1': ['K0', 'K0', 'K1', 'K2'],
                     'key2': ['K0', 'K1', 'K0', 'K1'],
                        'A': ['A0', 'A1', 'A2', 'A3'],
                        'B': ['B0', 'B1', 'B2', 'B3']})
    
right = pd.DataFrame({'key1': ['K0', 'K1', 'K1', 'K2'],
                               'key2': ['K0', 'K0', 'K0', 'K0'],
                                  'C': ['C0', 'C1', 'C2', 'C3'],
                                  'D': ['D0', 'D1', 'D2', 'D3']})
```

The default merge is **inner**: only rows whose (key1, key2) pair appears in **both** tables are kept. Notice that the pair (K1, K0) appears twice in `right`, so the matching row of `left` is repeated.

```python
pd.merge(left, right, on=['key1', 'key2'])
```

An **outer** merge keeps **every** row from both tables. Where one side has no match, its columns are filled with NaN.

```python
pd.merge(left, right, how='outer', on=['key1', 'key2'])
```

A **right** merge keeps every row of the `right` table, and adds data from `left` only where it matches.

```python
pd.merge(left, right, how='right', on=['key1', 'key2'])
```

A **left** merge keeps every row of the `left` table, and adds data from `right` only where it matches.

```python
pd.merge(left, right, how='left', on=['key1', 'key2'])
```

## Joining
Joining is a convenient method for combining the columns of two potentially differently-indexed DataFrames into a single result DataFrame.

Two new tables for joining. This time there is no key column: the rows are matched by their **index** (row labels). `left` has K0, K1, K2 and `right` has K0, K2, K3.

```python
left = pd.DataFrame({'A': ['A0', 'A1', 'A2'],
                     'B': ['B0', 'B1', 'B2']},
                      index=['K0', 'K1', 'K2']) 

right = pd.DataFrame({'C': ['C0', 'C2', 'C3'],
                    'D': ['D0', 'D2', 'D3']},
                      index=['K0', 'K2', 'K3'])
```

`left.join(right)` keeps all rows of `left` and adds the columns of `right` where the labels match. K1 is not in `right`, so it gets NaN.

```python
left.join(right)
```

With `how='outer'`, every label from both tables is kept (K0 to K3), with NaN where one side has no data.

```python
left.join(right, how='outer')
```
