# Operations

There are lots of operations with pandas that will be really useful to you, but don't fall into any distinct category. Let's show them here in this lecture:

## About this notebook

A collection of everyday pandas tools that you will use again and again: finding unique values, counting how often each value appears, filtering rows, applying your own function to a column, deleting columns, sorting, finding missing values, and building a pivot table.

**How to use it:** click a code cell and press **Shift + Enter** to run it. Run the cells from top to bottom, because later cells use things created in earlier ones. The short note above each code cell tells you what it does.

Import pandas, create a small example table and show it. `head()` displays the first five rows, which is the usual way to take a quick look at a table.

```python
import pandas as pd
df = pd.DataFrame({'col1':[1,2,3,4],'col2':[444,555,666,444],'col3':['abc','def','ghi','xyz']})
df.head()
```

### Info on Unique Values

`unique()` lists the **different** values in column `col2`, each one only once.

```python
df['col2'].unique()
```

`nunique()` counts **how many** different values there are.

```python
df['col2'].nunique()
```

`value_counts()` shows how many times each value appears. Here 444 appears twice.

```python
df['col2'].value_counts()
```

### Selecting Data

Filter rows using two columns at once: keep rows where `col1` is greater than 2 **and** `col2` equals 444. The result is stored in `newdf`.

```python
#Select from DataFrame using criteria from multiple columns
newdf = df[(df['col1']>2) & (df['col2']==444)]
```

Display the filtered rows.

```python
newdf
```

### Applying Functions

Define our own small function that doubles a number.

```python
def times2(x):
    return x*2
```

`apply()` runs a function on **every value** of a column. Here each value of `col1` is doubled.

```python
df['col1'].apply(times2)
```

`apply()` works with built-in functions too: `len` gives the length of each piece of text in `col3`.

```python
df['col3'].apply(len)
```

Add up all the values in `col1`.

```python
df['col1'].sum()
```

** Permanently Removing a Column**

`del` deletes the column `col1` **permanently** (in place, no copy is made).

```python
del df['col1']
```

The table now has only `col2` and `col3`.

```python
df
```

** Get column and index names: **

`df.columns` lists the column names.

```python
df.columns
```

`df.index` describes the row labels. `RangeIndex(start=0, stop=4, step=1)` simply means the rows are numbered 0, 1, 2, 3.

```python
df.index
```

** Sorting and Ordering a DataFrame:**

Display the table again before sorting.

```python
df
```

`sort_values(by='col2')` sorts the rows from the smallest to the largest value of `col2`. Notice that the row labels move along with their rows. (`df` itself is not changed.)

```python
df.sort_values(by='col2') #inplace=False by default
```

** Find Null Values or Check for Null Values**

`isnull()` checks every cell and shows **True** where a value is missing. This table has no gaps, so everything is False.

```python
df.isnull()
```

`dropna()` removes rows with missing values. There are none here, so the table comes back unchanged.

```python
# Drop rows with NaN Values
df.dropna()
```

** Filling in NaN values with something else: **

Import NumPy, which we need for `np.nan` (a missing value).

```python
import numpy as np
```

Create a new table that **does** contain missing values (NaN) and display it.

```python
df = pd.DataFrame({'col1':[1,2,3,np.nan],
                   'col2':[np.nan,555,666,444],
                   'col3':['abc','def','ghi','xyz']})
df.head()
```

`fillna('FILL')` replaces every missing value with the text `'FILL'`.

```python
df.fillna('FILL')
```

Create a new table for the pivot table example: columns A, B and C hold categories, and column D holds numbers.

```python
data = {'A':['foo','foo','foo','bar','bar','bar'],
     'B':['one','one','two','two','one','one'],
       'C':['x','y','x','y','x','y'],
       'D':[1,3,2,5,4,1]}

df = pd.DataFrame(data)
```

Display the table.

```python
df
```

A **pivot table** reorganises the data into a summary grid, as in Excel. Here the rows are the combinations of A and B, the columns are the values of C (x and y), and each cell shows the value of D (the average, if there were several). Combinations that do not exist in the data show **NaN**.

```python
df.pivot_table(values='D',index=['A', 'B'],columns=['C'])
```
