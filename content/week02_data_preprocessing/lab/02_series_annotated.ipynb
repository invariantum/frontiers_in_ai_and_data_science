
# Series

## About this notebook

A **Series** is the simplest pandas object: a list of values where every value has a **label** (called the *index*). Here you will learn how to create a Series from a list, a NumPy array and a dictionary, how to look up a value by its label, and what happens when you add two Series together.

**How to use it:** click a code cell and press **Shift + Enter** to run it. Run the cells from top to bottom, because later cells use things created in earlier ones. The short note above each code cell tells you what it does.

The first main data type we will learn about for pandas is the Series data type. Let's import Pandas and explore the Series object.

A Series is very similar to a NumPy array (in fact it is built on top of the NumPy array object). What differentiates the NumPy array from a Series, is that a Series can have axis labels, meaning it can be indexed by a label, instead of just a number location. It also doesn't need to hold numeric data, it can hold any arbitrary Python Object.

Let's explore this concept through some examples:

First we **import** the two libraries we need. By convention, NumPy is shortened to `np` and pandas to `pd`, so we can type `pd.Series` instead of `pandas.Series`.

```python
import numpy as np
import pandas as pd
```

### Creating a Series

You can convert a list,numpy array, or dictionary to a Series:

Here we prepare some sample data to build Series from: a list of labels, a list of numbers, a NumPy array with the same numbers, and a dictionary that pairs each label with a number. Nothing is displayed, the values are just stored.

```python
labels = ['a','b','c']
my_list = [10,20,30]
arr = np.array([10,20,30])
d = {'a':10,'b':20,'c':30}
```

** Using Lists**

Build a Series from a plain list. Because we gave no labels, pandas numbers the rows automatically: **0, 1, 2**. The left column is the index (labels), the right column is the data.

```python
pd.Series(data=my_list)
```

Same list, but now we give our own labels with `index=labels`, so the rows are called **a, b, c** instead of 0, 1, 2.

```python
pd.Series(data=my_list,index=labels)
```

This is exactly the same as the previous cell. The first two inputs of `pd.Series` are the data and the index, so we can leave out the names `data=` and `index=`.

```python
pd.Series(my_list,labels)
```

** NumPy Arrays **

A Series can also be built from a NumPy array. The result looks the same as with a list.

```python
pd.Series(arr)
```

A NumPy array plus our own labels.

```python
pd.Series(arr,labels)
```

** Dictionary**

When you build a Series from a **dictionary**, pandas uses the dictionary's keys as the labels and its values as the data, so you do not need to give an index.

```python
pd.Series(d)
```

### Data in a Series

A pandas Series can hold a variety of object types:

A Series does not have to hold numbers. Here it holds text (strings). The `dtype` line at the bottom tells you what kind of values the Series holds: `str` means text.

```python
pd.Series(data=labels)
```

Just to show how flexible a Series is: it can even hold Python functions (`dtype: object` means "any kind of Python object"). You will rarely need this in practice.

```python
# Even functions (although unlikely that you will use this)
pd.Series([sum,print,len])
```

## Using an Index

The key to using a Series is understanding its index. Pandas makes use of these index names or numbers by allowing for fast look ups of information (works like a hash table or dictionary).

Let's see some examples of how to grab information from a Series. Let us create two sereis, ser1 and ser2:

Create a Series of four numbers labelled with country names and store it in the variable `ser1`.

```python
ser1 = pd.Series([1,2,3,4],index = ['USA', 'Germany','USSR', 'Japan'])                                   
```

Typing the name of a variable on the last line of a cell displays it.

```python
ser1
```

A second Series, `ser2`. Note that it has **Italy** instead of **USSR**, so the two Series do not have exactly the same labels.

```python
ser2 = pd.Series([1,2,5,4],index = ['USA', 'Germany','Italy', 'Japan'])                                   
```

Display `ser2`.

```python
ser2
```

Look up a single value **by its label**, just like looking up a key in a dictionary.

```python
ser1['USA']
```

Operations are then also done based off of index:

When you add two Series, pandas matches the values **by label**, not by position. USA is added to USA, Japan to Japan, and so on. **Italy** and **USSR** appear in only one of the two Series, so there is nothing to add them to and the result is **NaN** (a missing value). The numbers also become decimals (`float64`), because NaN can only be stored that way.

```python
ser1 + ser2
```
