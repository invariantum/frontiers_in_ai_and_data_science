## About this notebook

This short page introduces **pandas**, the Python library we use this week to load, clean and organise data before analysing it (this is called *data pre-processing*). It contains no code: it simply lists the notebooks of this lab in the order you should work through them.

**Two words you will see everywhere:**
* **Series**: a single column of values, each with a label.
* **DataFrame**: a table made of several Series that share the same row labels, like a sheet in Excel.
