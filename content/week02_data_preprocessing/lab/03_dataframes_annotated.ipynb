# DataFrames

DataFrames are the workhorse of pandas and are directly inspired by the R programming language. We can think of a DataFrame as a bunch of Series objects put together to share the same index. Let's use pandas to explore this topic!

## About this notebook

A **DataFrame** is a table: rows have labels (the *index*) and columns have names. It is the object you will use most in data science. Here you will learn how to select columns and rows, add and remove columns, filter rows with conditions, change the index, and work with a two-level (*multi*) index.

**How to use it:** click a code cell and press **Shift + Enter** to run it. Run the cells from top to bottom, because later cells use things created in earlier ones. The short note above each code cell tells you what it does.

Import pandas and NumPy, as in every pandas notebook.

```python
import pandas as pd
import numpy as np
```

`randn` creates random numbers. Setting the **seed** to 101 makes the "random" numbers the same every time you run the notebook, so your results match everyone else's.

```python
from numpy.random import randn
np.random.seed(101)
```

Create a DataFrame of 5 rows by 4 columns filled with random numbers. The rows are labelled **A to E** and the columns **W, X, Y, Z**. (`'A B C D E'.split()` is a quick way to write the list `['A','B','C','D','E']`.)

```python
df = pd.DataFrame(randn(5,4),index='A B C D E'.split(),columns='W X Y Z'.split())
```

Display the table.

```python
df
```

## Selection and Indexing

Let's learn the various methods to grab data from a DataFrame

Select one column by putting its name in square brackets. A single column comes back as a **Series**.

```python
df['W']
```

To select several columns, pass a **list** of names. Note the double brackets: the outer ones select, the inner ones make the list. The result is a smaller DataFrame.

```python
# Pass a list of column names
df[['W','Z']]
```

You can also select a column with a dot, like `df.W`. It works, but it is not recommended: it fails if the column name contains a space or has the same name as a pandas method. Prefer `df['W']`.

```python
# SQL Syntax (NOT RECOMMENDED!)
df.W
```

DataFrame Columns are just Series

Check the type of a single column: it is a pandas **Series**. A DataFrame is really a set of Series that share the same row labels.

```python
type(df['W'])
```

**Creating a new column:**

Create a **new column** called `new` by adding columns W and Y, row by row.

```python
df['new'] = df['W'] + df['Y']
```

The table now has a fifth column, `new`.

```python
df
```

** Removing Columns**

Remove the column `new`. `axis=1` means "look for a **column** with this name" (`axis=0` would mean a row). The result is shown without `new`...

```python
df.drop('new',axis=1)
```

...but the original table has **not** changed: `new` is still there. Most pandas methods return a new copy and leave the original untouched.

```python
# Not inplace unless specified!
df
```

To change the original table itself, add `inplace=True`. This cell shows nothing because it modifies `df` directly.

```python
df.drop('new',axis=1,inplace=True)
```

Now the column `new` is really gone.

```python
df
```

Can also drop rows this way:

Drop a **row** instead, by using its label and `axis=0`. (Again this is not in place, so `df` still has row E.)

```python
df.drop('E',axis=0)
```

** Selecting Rows**

`.loc[...]` selects rows **by label**. Here we get row A, returned as a Series.

```python
df.loc['A']
```

Or select based off of position instead of label 

`.iloc[...]` selects rows **by position**, counting from 0. Position 2 is the third row, which is row C.

```python
df.iloc[2]
```

** Selecting subset of rows and columns **

With `.loc[row, column]` you can pick a single value: the one in row B, column Y.

```python
df.loc['B','Y']
```

Pass lists to get a smaller block of the table: rows A and B, columns W and Y.

```python
df.loc[['A','B'],['W','Y']]
```

### Conditional Selection

An important feature of pandas is conditional selection using bracket notation, very similar to numpy:

Display the table again, as a reminder before we start filtering.

```python
df
```

Comparing the whole table with a number gives a table of **True/False** values: True wherever the value is greater than 0.

```python
df>0
```

Using that True/False table inside the brackets keeps the values where it is True and replaces the rest with **NaN**.

```python
df[df>0]
```

More useful in practice: keep only the **rows** where column W is greater than 0. Rows with a negative W disappear.

```python
df[df['W']>0]
```

You can chain selections: first filter the rows where W > 0, then take column Y from the result.

```python
df[df['W']>0]['Y']
```

Same idea, but take two columns, Y and X, from the filtered rows.

```python
df[df['W']>0][['Y','X']]
```

For two conditions you can use | and & with parenthesis:

Combine two conditions with **`&`** (and) or **`|`** (or). Each condition must be inside its own parentheses. Here we keep rows where W > 0 **and** Y > 1. (Python's words `and`/`or` do not work here.)

```python
df[(df['W']>0) & (df['Y'] > 1)]
```

## More Index Details

Let's discuss some more features of indexing, including resetting the index or setting it something else. We'll also talk about index hierarchy!

Display the table again before changing its index.

```python
df
```

`reset_index()` turns the row labels (A to E) into an ordinary column called `index`, and numbers the rows 0, 1, 2... instead. Not in place, so `df` itself is unchanged.

```python
# Reset to default 0,1...n index
df.reset_index()
```

A quick way to make the list `['CA', 'NY', 'WY', 'OR', 'CO']` of US state codes.

```python
newind = 'CA NY WY OR CO'.split()
```

Add that list to the table as a new column called `States` (one value per row).

```python
df['States'] = newind
```

The table now has a `States` column.

```python
df
```

`set_index('States')` uses the `States` column as the row labels instead of A to E. Again, this only shows the result...

```python
df.set_index('States')
```

...and `df` is unchanged.

```python
df
```

With `inplace=True` the change is made to `df` itself.

```python
df.set_index('States',inplace=True)
```

Now the rows are labelled by state.

```python
df
```

## Multi-Index and Index Hierarchy

Let us go over how to work with Multi-Index, first we'll create a quick example of what a Multi-Indexed DataFrame would look like:

Now we build a **multi-index**: row labels with two levels, like chapters and sections. The outer level is the group (G1, G2) and the inner level is a number (1, 2, 3). `zip` pairs them up as (G1, 1), (G1, 2) and so on, and `MultiIndex.from_tuples` turns those pairs into an index.

```python
# Index Levels
outside = ['G1','G1','G1','G2','G2','G2']
inside = [1,2,3,1,2,3]
hier_index = list(zip(outside,inside))
hier_index = pd.MultiIndex.from_tuples(hier_index)
```

Display the multi-index we just built: six pairs of labels.

```python
hier_index
```

Create a DataFrame of random numbers with two columns (A and B) that uses the two-level index for its rows.

```python
df = pd.DataFrame(np.random.randn(6,2),index=hier_index,columns=['A','B'])
df
```

Now let's show how to index this! For index hierarchy we use df.loc[], if this was on the columns axis, you would just use normal bracket notation df[]. Calling one level of the index returns the sub-dataframe:

`.loc['G1']` selects the **outer** level: all rows that belong to group G1.

```python
df.loc['G1']
```

Go one level deeper: from group G1, take the row with inner label 1. The result is a single row (a Series).

```python
df.loc['G1'].loc[1]
```

`index.names` shows the names of the two index levels. They are `None` because we have not named them yet.

```python
df.index.names
```

Give the two levels names: **Group** for the outer level and **Num** for the inner one.

```python
df.index.names = ['Group','Num']
```

The level names now appear above the row labels.

```python
df
```

`.xs()` (short for *cross-section*) is another way to select from a multi-index. `xs('G1')` gives all rows of group G1, the same as `.loc['G1']`.

```python
df.xs('G1')
```

Select one exact row by giving both labels as a pair (a **tuple**): group G1, number 1.

```python
df.xs(('G1',1))
```

This is where `xs` is really handy: it can select from the **inner** level. Here we take the row with Num = 1 from **every** group, which is hard to do with `.loc`.

```python
df.xs(1,level='Num')
```
