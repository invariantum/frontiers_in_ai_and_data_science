## Data types

Every value in Python has a **type**, which determines what you can do with it. You can always check a value's type with `type(value)`.

### Numbers

Python has two main numeric types:
* `int`: whole numbers, e.g. `2`, `-7`, `1000`
* `float`: numbers with a decimal point, e.g. `0.5`, `3.14`

The basic arithmetic operators are:

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` | addition | `1 + 1` | `2` |
| `*` | multiplication | `1 * 3` | `3` |
| `/` | division (always returns a float) | `1 / 2` | `0.5` |
| `**` | power (exponent) | `2 ** 4` | `16` |
| `%` | modulo (remainder of a division) | `5 % 2` | `1` |

````python
# Addition of two integers
1 + 1
````

````python
# Multiplication
1 * 3
````

````python
# Division: in Python 3, "/" always returns a float, even if both numbers are integers.
# (Use "//" if you want integer "floor" division: 7 // 2 gives 3.)
1 / 2
````

````python
# Exponentiation: 2 to the power of 4 = 2 * 2 * 2 * 2
# Note: Python uses "**" for powers. The "^" symbol means something else (bitwise XOR).
2 ** 4
````

````python
# Modulo returns the REMAINDER of a division.
# 4 divided by 2 is exactly 2, so the remainder is 0.
4 % 2
````

````python
# 5 divided by 2 is 2 with remainder 1.
# A common use: "number % 2 == 0" tests whether a number is even.
5 % 2
````

````python
# Parentheses control the order of operations, exactly as in mathematics.
# First (2 + 3) = 5 and (5 + 5) = 10 are evaluated, then 5 * 10 = 50.
(2 + 3) * (5 + 5)
````

### Variable assignment

A **variable** is a name that refers to a value stored in memory. The `=` sign means *assign* (store the value on the right under the name on the left); it is not a mathematical equality.

Naming rules:
* Names can contain letters, digits and underscores (`_`), but **cannot start with a digit**.
* They cannot contain spaces or special characters such as `-`, `@`, `!`.
* Names are case-sensitive: `Age` and `age` are different variables.
* By convention, Python uses lowercase words separated by underscores (`snake_case`), e.g. `total_sales`.

````python
# Can not start with number or special characters
# Valid: name_of_var, x1, _temp      Invalid: 1var, my-var, @x
name_of_var = 2
````

````python
# Assign the value 2 to x and the value 3 to y.
# Assignment produces no output; the values are simply stored in memory.
x = 2
y = 3
````

````python
# Python first evaluates the right-hand side (x + y = 2 + 3 = 5),
# then stores the result in a new variable called z.
z = x + y
````

````python
# Typing a variable name as the last line of a cell displays its value.
z
````

### Strings

A **string** (`str`) is a sequence of characters, i.e. text. Strings can be written with single quotes `'...'` or double quotes `"..."`; both are equivalent.

Having two kinds of quotes is useful when the text itself contains a quote character.

````python
# A string defined with single quotes
'single quotes'
````

````python
# The same kind of string defined with double quotes.
# Notice that Jupyter displays it with single quotes; the quote style is not part of the value.
"double quotes"
````

````python
# Using double quotes lets us include an apostrophe (a single quote) inside the text.
# With single quotes, 'lot's' would end the string too early and cause a SyntaxError.
" wrap lot's of other quotes"
````

### Printing

There are two ways to see a value in a notebook:
* **Displaying** it, by putting it on the last line of a cell. Jupyter shows its *representation* (for strings, with quotes).
* **Printing** it with `print()`. This shows the *content* in a human-friendly way (no quotes) and works anywhere, including inside loops and functions and in normal Python scripts.

To build a sentence that contains variables, we use **string formatting** with the `.format()` method.

````python
# Store the text 'hello' in a variable called x (this replaces the earlier value x = 2).
x = 'hello'
````

````python
# Displaying the variable: the output shows the quotes, i.e. the "representation" of the string.
x
````

````python
# Printing the variable: the output shows only the text, without quotes.
print(x)
````

````python
# Two variables of different types: an integer and a string.
num = 12
name = 'Sam'
````

````python
# .format() replaces each {placeholder} with a value.
# Here the placeholders are NAMED ({one} and {two}), and we tell .format() which value goes where.
# Named placeholders make long templates easier to read.
print('My number is: {one}, and my name is: {two}'.format(one=num,two=name))
````

````python
# Empty placeholders {} are filled by POSITION: the first {} gets num, the second gets name.
# .format() converts the number 12 to text automatically.
print('My number is: {}, and my name is: {}'.format(num,name))
````

````python
# Modern alternative (Python 3.6+): an "f-string".
# Putting f before the quotes lets you write variable names directly inside the braces.
# You will see this style very often in data science code.
print(f'My number is: {num}, and my name is: {name}')
````

### Lists

A **list** is an *ordered* and *mutable* (changeable) collection of items, written with square brackets `[ ]` and separated by commas.

* Lists can hold items of **any type**, even mixed types or other lists.
* Each item has a position called an **index**. Python counts from **0**, so the first item is at index `0`.
* **Slicing** with `list[start:stop]` returns a sub-list from `start` up to, but **not including**, `stop`.

Lists are one of the most-used structures in Python. In data science, a column of data is often first loaded or built as a list.

````python
# A list of three integers
[1,2,3]
````

````python
# A list can mix types: a string, an integer, and another list (a "nested" list).
['hi',1,[1,2]]
````

````python
# Store a list of three strings in a variable.
my_list = ['a','b','c']
````

````python
# .append() adds an item to the END of the list.
# It changes the list "in place" (the original list is modified) and returns nothing.
my_list.append('d')
````

````python
# The list now has four items.
my_list
````

````python
# Indexing: get the item at position 0, i.e. the FIRST item.
my_list[0]
````

````python
# Index 1 is the SECOND item.
# Tip: negative indices count from the end, so my_list[-1] is the last item.
my_list[1]
````

````python
# Slicing: from index 1 to the end of the list.
# Leaving out the stop value means "go all the way to the end".
my_list[1:]
````

````python
# Slicing: from the start up to, but NOT including, index 1.
# The result is still a list (with a single item), not the item itself.
my_list[:1]
````

````python
# Lists are MUTABLE: we can replace an item by assigning to its index.
my_list[0] = 'NEW'
````

````python
# The first item has been replaced.
my_list
````

**Nested lists.** When a list contains other lists, we chain indices to "drill down" one level at a time, reading from left to right.

````python
# A list whose 4th item (index 3) is itself a list, which in turn contains another list.
nest = [1,2,3,[4,5,['target']]]
````

````python
# Level 1: index 3 of nest is the inner list [4, 5, ['target']].
nest[3]
````

````python
# Level 2: index 2 of that inner list is ['target'] (still a list).
nest[3][2]
````

````python
# Level 3: index 0 of ['target'] is the string 'target' itself.
nest[3][2][0]
````

### Dictionaries

A **dictionary** (`dict`) stores data as **key: value** pairs, written with curly braces `{ }`.

* Instead of looking items up by position (as in a list), you look them up by their **key**.
* Keys must be unique. Values can be anything, including lists or other dictionaries.
* Dictionaries are like a real dictionary: you look up a word (the key) to get its definition (the value).

Dictionaries are central in data science. For example, a single data record such as `{'name': 'Jones', 'salary': 25000}`, or JSON data from a web API, maps naturally onto a dictionary.

````python
# A dictionary with two key-value pairs.
d = {'key1':'item1','key2':'item2'}
````

````python
# Display the whole dictionary.
d
````

````python
# Look up the value stored under the key 'key1'.
# If the key does not exist, Python raises a KeyError.
d['key1']
````

### Booleans

A **Boolean** (`bool`) has only two possible values: `True` or `False` (note the capital letters).

Booleans are the result of comparisons and are used to make decisions in `if` statements and loops. In data science they are used constantly to **filter** data, for example to keep only rows where `age > 30`.

````python
# The Boolean value True
True
````

````python
# The Boolean value False
False
````

### Tuples

A **tuple** is an ordered collection, like a list, but written with parentheses `( )`. The key difference is that tuples are **immutable**: once created, their items cannot be changed.

Use a tuple for data that should not change, for example a coordinate pair `(latitude, longitude)` or the shape of a data table `(rows, columns)`.

````python
# Create a tuple with three items.
t = (1,2,3)
````

````python
# Indexing works exactly as with lists.
t[0]
````

````python
# Trying to change an item FAILS on purpose, because tuples are immutable.
# The resulting TypeError is the expected outcome of this cell; it demonstrates the rule.
t[0] = 'NEW'
````

### Sets

A **set** is an *unordered* collection of **unique** items, written with curly braces `{ }` but without key-value pairs.

Sets automatically remove duplicates, which makes them very handy for finding the distinct values in a dataset (for example, all the different product categories in a sales table).

````python
# A set of three items
{1,2,3}
````

````python
# Duplicates are discarded automatically: only the unique values 1, 2 and 3 remain.
{1,2,3,1,2,1,2,3,3,3,3,2,2,2,1,1,2}
````

## Comparison operators

Comparison operators compare two values and always return a Boolean (`True` or `False`).

| Operator | Meaning |
|---|---|
| `>` | greater than |
| `<` | less than |
| `>=` | greater than or equal to |
| `<=` | less than or equal to |
| `==` | equal to (note: **two** equals signs; a single `=` is assignment) |
| `!=` | not equal to |

````python
# Is 1 greater than 2? No.
1 > 2
````

````python
# Is 1 less than 2? Yes.
1 < 2
````

````python
# "Greater than OR equal to" is True when the values are equal.
1 >= 1
````

````python
# Is 1 less than or equal to 4? Yes.
1 <= 4
````

````python
# Equality test uses "==".
1 == 1
````

````python
# Comparisons also work on strings: these two strings are not identical, so the result is False.
'hi' == 'bye'
````

## Logic operators

Logic operators combine several Boolean conditions into one:

* `and`: True only if **both** conditions are True.
* `or`: True if **at least one** condition is True.
* `not`: reverses a Boolean (`not True` is `False`).

The parentheses are optional here, but they make the code much easier to read.

````python
# (1 > 2) is False and (2 < 3) is True.
# With "and", a single False makes the whole expression False.
(1 > 2) and (2 < 3)
````

````python
# With "or", one True condition is enough to make the whole expression True.
(1 > 2) or (2 < 3)
````

````python
# Several conditions chained with "or": False or False or True gives True.
(1 == 2) or (2 == 3) or (4 == 4)
````

## `if`, `elif`, `else` statements

Conditional statements let a program **make decisions**: run a block of code only if a condition is True.

```python
if condition_1:
    # runs if condition_1 is True
elif condition_2:
    # runs if condition_1 is False AND condition_2 is True
else:
    # runs if none of the conditions above were True
```

Two important syntax rules:
1. Each condition line ends with a colon `:`.
2. The code that belongs to a branch is **indented** (4 spaces by convention). In Python, indentation is not just style; it defines which lines belong to the block.

Python checks the conditions from top to bottom and runs **only the first** branch whose condition is True.

````python
# The condition 1 < 2 is True, so the indented line runs.
if 1 < 2:
    print('Yep!')
````

````python
# Same logic. The point: everything indented under "if" belongs to that block.
if 1 < 2:
    print('yep!')
````

````python
# An if/else gives two alternatives. The condition is True, so the "if" branch runs.
if 1 < 2:
    print('first')
else:
    print('last')
````

````python
# Now the condition is False, so Python skips the "if" block and runs the "else" block.
if 1 > 2:
    print('first')
else:
    print('last')
````

````python
# Python checks conditions in order:
#   1 == 2  -> False, skip
#   3 == 3  -> True, run this branch and then STOP checking
# The "else" branch is never reached.
if 1 == 2:
    print('first')
elif 3 == 3:
    print('middle')
else:
    print('Last')
````

## `for` loops

A `for` loop **repeats** a block of code once for every item in a sequence (a list, a string, a range, and so on).

```python
for item in sequence:
    # this block runs once per item; "item" holds the current element
```

The loop variable name (`item` below) is chosen by you. It can be any valid name; it simply refers to the current element on each pass.

````python
# A list to loop over
seq = [1,2,3,4,5]
````

````python
# On each pass, "item" takes the next value from seq (1, then 2, ... then 5), and we print it.
for item in seq:
    print(item)
````

````python
# The loop body does not have to use the loop variable.
# Here we print 'Yep' once for each of the 5 items in the list.
for item in seq:
    print('Yep')
````

````python
# The loop variable can have ANY name; "jelly" works just as well as "item".
# (In practice, choose meaningful names.) Each value is added to itself, i.e. doubled.
for jelly in seq:
    print(jelly+jelly)
````

## `while` loops

A `while` loop keeps repeating **as long as a condition remains True**. It is useful when you do not know in advance how many repetitions you need.

**Warning:** something inside the loop must eventually make the condition False. Otherwise you create an **infinite loop**. (If that happens in Jupyter, use the *Interrupt / Stop* button.)

````python
i = 1                               # start the counter at 1
while i < 5:                        # keep looping while i is less than 5
    print('i is: {}'.format(i))     # show the current value of i
    i = i+1                         # increase i by 1; without this line the loop would never end
# When i becomes 5, the condition 5 < 5 is False and the loop stops, so 5 is never printed.
````

## `range()`

`range()` generates a sequence of integers. It is most often used with `for` loops to repeat something a fixed number of times.

* `range(stop)`: from 0 up to, but not including, `stop`
* `range(start, stop)`: from `start` up to, but not including, `stop`
* `range(start, stop, step)`: counts in steps, e.g. `range(0, 10, 2)` gives 0, 2, 4, 6, 8

````python
# range is "lazy": it does not create all the numbers immediately, it produces them when needed.
# That is why displaying it just shows range(0, 5) instead of the individual numbers.
range(5)
````

````python
# Looping over range(5) gives the numbers 0, 1, 2, 3, 4 (five numbers; 5 itself is excluded).
for i in range(5):
    print(i)
````

````python
# Converting a range to a list makes all its numbers visible at once.
list(range(5))
````

## List comprehension

A **list comprehension** is a compact, one-line way to build a new list by transforming each item of an existing sequence.

```python
[expression for item in sequence]
```

Below, the same task (squaring every number) is done twice: first with a traditional `for` loop, then with a list comprehension. Both give exactly the same result. List comprehensions are very common in data science code because they are short and readable.

````python
# The input list
x = [1,2,3,4]
````

````python
# Approach 1: a traditional loop
out = []                    # 1. start with an empty list
for item in x:              # 2. go through each number in x
    out.append(item**2)     # 3. square it and add it to the end of "out"
print(out)
````

````python
# Approach 2: the same result in ONE line with a list comprehension.
# Read it as: "item squared, for each item in x".
[item**2 for item in x]
````

## Functions

A **function** is a named, reusable block of code. You define it once and can then *call* it as many times as you like. Functions help avoid repetition and keep programs organised.

```python
def function_name(parameter):
    """Docstring: a short description of what the function does."""
    # body
    return result
```

Key ideas:
* `def` starts the definition; the body is indented.
* **Parameters** are the inputs. A parameter can have a **default value**, which is used when no argument is given.
* A **docstring** (the text in triple quotes) documents the function. You can read it with `help(function_name)`.
* `return` sends a value back to the caller. A function without `return` returns `None`.
* **`print` is not the same as `return`**: `print` only shows something on screen, while `return` gives back a value that you can store and use later.

````python
# Define a function with ONE parameter, param1, whose default value is 'default'.
# Defining a function does not run it; it only stores it so we can use it later.
def my_func(param1='default'):
    """
    Docstring goes here.
    """
    print(param1)   # the function prints its input; it does not return anything
````

````python
# Without parentheses we refer to the function OBJECT itself; we do not call it.
# The output just confirms that my_func is a function.
my_func
````

````python
# Calling the function WITHOUT an argument uses the default value.
my_func()
````

````python
# Calling it with a POSITIONAL argument: 'new param' replaces the default.
my_func('new param')
````

````python
# Calling it with a KEYWORD argument: we name the parameter explicitly.
# This is clearer when a function has many parameters (very common in data science libraries).
my_func(param1='new param')
````

````python
# A function that RETURNS a value instead of printing it.
def square(x):
    return x**2
````

````python
# The returned value (2 squared = 4) is stored in the variable "out".
# Nothing is displayed, because assignment produces no output.
out = square(2)
````

````python
# Now we can use the stored result.
print(out)
````

## Lambda expressions

A **lambda expression** is a small, anonymous (unnamed) function written in a single line:

```python
lambda arguments: expression
```

The value of the expression is returned automatically, without a `return` keyword. Lambdas are useful for short, throwaway functions that are passed to other functions, such as `map()`, `filter()` or `sorted()`, and later in pandas (e.g. `df['col'].apply(lambda v: v * 2)`).

The two cells below define the same behaviour in both styles.

````python
# A normal function that doubles its input
def times2(var):
    return var*2
````

````python
# Calling it: 2 * 2 = 4
times2(2)
````

````python
# The equivalent lambda: takes "var" and returns var*2.
# On its own it just creates a function object; we have not called it or stored it.
lambda var: var*2
````

## `map` and `filter`

These two built-in functions apply a function to **every item** of a sequence:

* `map(function, sequence)`: **transforms** each item and returns the results.
* `filter(function, sequence)`: **keeps only** the items for which the function returns `True`.

Both are "lazy", like `range`: they return an object that produces results on demand. Wrap them in `list()` to see all the results at once.

````python
# The input sequence
seq = [1,2,3,4,5]
````

````python
# map applies times2 to each item, but displaying it only shows a "map object" (and its memory address).
map(times2,seq)
````

````python
# Converting to a list forces the computation: every number is doubled.
list(map(times2,seq))
````

````python
# The same transformation using a lambda instead of a named function.
list(map(lambda var: var*2,seq))
````

````python
# filter keeps the items for which the function returns True.
# Again, displaying it directly shows only a "filter object".
filter(lambda item: item%2 == 0,seq)
````

````python
# item % 2 == 0 is True for even numbers, so only 2 and 4 are kept.
list(filter(lambda item: item%2 == 0,seq))
````

## Methods

A **method** is a function that belongs to a particular type of object. It is called with **dot notation**: `object.method()`. Different types have different methods: strings have text methods, lists have list methods, and so on.

Tip: in Jupyter, type an object's name followed by a dot and press **Tab** to see all its available methods.

````python
# A sample sentence
st = 'hello my name is Sam'
````

````python
# .lower() returns a NEW string in lowercase.
# Strings are immutable, so the original "st" is not changed.
st.lower()
````

````python
# .upper() returns a new string in uppercase.
st.upper()
````

````python
# .split() with no argument splits the string on whitespace and returns a LIST of words.
# This is the first step of many text-processing and NLP tasks ("tokenisation").
st.split()
````

````python
# A short social-media-style text with a hashtag
tweet = 'Go Sports! #Sports'
````

````python
# .split('#') splits the text at every '#' character instead of at spaces.
# The result has two parts: the text before the '#' and the text after it.
tweet.split('#')
````

````python
# Combine a method call with indexing: take the part at index 1, i.e. the hashtag word.
tweet.split('#')[1]
````

````python
# Reminder: the dictionary we created earlier
d
````

````python
# .keys() returns all the keys of the dictionary.
d.keys()
````

````python
# .items() returns all (key, value) pairs as tuples. This is very useful for looping:
#     for key, value in d.items(): ...
d.items()
````

````python
# A new list
lst = [1,2,3]
````

````python
# .pop() REMOVES the last item from the list and RETURNS it (so we see 3).
# lst.pop(0) would remove and return the first item instead.
lst.pop()
````

````python
# The list has been modified in place: 3 is gone.
lst
````

````python
# The "in" operator tests MEMBERSHIP: is 'x' an item of this list? No.
'x' in [1,2,3]
````

````python
# 'x' is an item of this list, so the result is True.
'x' in ['x','y','z']
````
