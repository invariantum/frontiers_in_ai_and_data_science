**What is the main difference between a tuple and a list?**

```python
# Answer: a tuple is IMMUTABLE, while a list is MUTABLE.
#
# - A list, e.g. [1, 2, 3], can be changed after creation: items can be replaced,
#   added (.append) or removed (.pop).
# - A tuple, e.g. (1, 2, 3), cannot be changed once created: t[0] = 'NEW' raises a TypeError.
#
# Practical consequence: use a tuple for fixed data that should not change by accident
# (e.g. coordinates, or the shape of a table), and a list for collections that grow or change.
```

**Create a function that grabs the email website domain from a string in the form:**

    user@domain.com

**So, for example, passing "user@domain.com" would return: domain.com**

```python
def domainGet(email):
    # email.split('@') breaks the string at the '@' character:
    #   'user@domain.com'.split('@')  ->  ['user', 'domain.com']
    # [-1] selects the LAST element of that list, i.e. the part after '@'.
    # (Using -1 rather than 1 is a safe choice: it always picks the final piece.)
    return email.split('@')[-1]
```

```python
# Test the function (run your function definition first).
# Expected output: 'domain.com'
domainGet('user@domain.com')
```

**Create a basic function that returns True if the word 'dog' is contained in the input string. Don't worry about edge cases like punctuation being attached to the word dog, but do account for capitalization.**

```python
def findDog(st):
    # 1. st.lower()  -> makes the check case-insensitive ('Dog' and 'DOG' become 'dog')
    # 2. .split()    -> turns the sentence into a list of words
    # 3. 'dog' in    -> membership test, which returns True or False
    # The comparison itself already produces a Boolean, so we can return it directly.
    return 'dog' in st.lower().split()
```

```python
# Test the function.
# Expected output: True
findDog('Is there a dog here?')
```

```python
# Edge cases (ignored by the task, but useful to understand the limits of the solution):
# 'Dog' is found thanks to .lower(), but 'dog?' is NOT equal to 'dog',
# because the punctuation stays attached to the word after .split().
print(findDog('My Dog is friendly'))   # True
print(findDog('Where is my dog?'))     # False: the word is 'dog?', not 'dog'
```

**Create a function that counts the number of times the word "dog" occurs in a string. Again, ignore edge cases.**

```python
def countDog(st):
    count = 0                          # 1. start the counter at zero
    for word in st.lower().split():    # 2. loop over every word (lowercased, for case-insensitivity)
        if word == 'dog':              # 3. check whether the current word is exactly 'dog'
            count += 1                 #    if so, add 1 (shorthand for count = count + 1)
    return count                       # 4. after the loop, return the total
```

```python
# Test the function.
# Expected output: 2
countDog('This dog runs faster than the other dog dude!')
```

```python
# Alternative one-line solution: a list's .count() method counts how often a value appears.
'This dog runs faster than the other dog dude!'.lower().split().count('dog')
```

**Use lambda expressions and the filter() function to filter out words from a list that don't start with the letter 's'. For example:**

    seq = ['soup','dog','salad','cat','great']

**should be filtered down to:**

    ['soup','salad']

```python
# The list of words to filter (run this cell first).
seq = ['soup','dog','salad','cat','great']
```

```python
# lambda word: word[0]=='s'  -> a small anonymous function; returns True if the first letter is 's'
# filter(..., seq)           -> applies it to each word and keeps only those that returned True
# list(...)                  -> filter is lazy, so we convert its result to a list to display it
list(filter(lambda word: word[0]=='s',seq))
```

```python
# Alternative: the string method .startswith() is more readable,
# and it also works safely on empty strings (where word[0] would raise an IndexError).
list(filter(lambda word: word.startswith('s'), seq))
```

**You are driving a little too fast, and a police officer stops you. Write a function to return one of 3 possible results: "No Ticket", "Small Ticket", or "Big Ticket".
If your speed is 60 or less, the result is "No Ticket". If speed is between 61 and 80 inclusive, the result is "Small Ticket". If speed is 81 or more, the result is "Big Ticket". Unless it is your birthday (encoded as a Boolean value in the parameters of the function): on your birthday, your speed can be 5 higher in all cases.**

```python
def caught_speeding(speed, is_birthday):

    # Step 1: handle the birthday rule.
    # On your birthday every limit is 5 higher. Instead of changing all three limits,
    # we subtract 5 from the speed and then compare it with the normal limits; the effect is the same.
    if is_birthday:
        speeding = speed - 5
    else:
        speeding = speed

    # Step 2: check the bands from the highest down.
    # Because "elif" is only reached when the previous condition was False,
    # "speeding > 60" here effectively means 61 to 80 inclusive.
    if speeding > 80:
        return 'Big Ticket'       # 81 or more
    elif speeding > 60:
        return 'Small Ticket'     # 61 to 80
    else:
        return 'No Ticket'        # 60 or less
```

```python
# Birthday: 81 - 5 = 76, which falls in the 61-80 band.
# Expected output: 'Small Ticket'
caught_speeding(81,True)
```

```python
# Not a birthday: 81 is above 80.
# Expected output: 'Big Ticket'
caught_speeding(81,False)
```

```python
# Extra checks at the boundaries: good practice whenever you write conditions.
print(caught_speeding(60, False))   # No Ticket    (exactly at the limit)
print(caught_speeding(61, False))   # Small Ticket
print(caught_speeding(80, False))   # Small Ticket
print(caught_speeding(65, True))    # No Ticket    (65 - 5 = 60)
print(caught_speeding(85, True))    # Small Ticket (85 - 5 = 80)
print(caught_speeding(86, True))    # Big Ticket   (86 - 5 = 81)
```
