```python
import importlib

packages = [
    "numpy", "pandas", "scipy", "sklearn", "statsmodels",
    "matplotlib", "seaborn", "plotly", "ipywidgets",
    "mlxtend", "sympy", "networkx", "PIL",
    "openpyxl", "xlrd", "tabulate",
]

for name in packages:
    try:
        module = importlib.import_module(name)
        version = getattr(module, "__version__", "?")
        print(f"OK      {name:<12} {version}")
    except Exception as e:
        print(f"FAILED  {name:<12} {type(e).__name__}: {e}")
```

```python
# Can pandas write and read an Excel file? (tests the pandas + openpyxl pairing)
import io
import pandas as pd

try:
    buffer = io.BytesIO()
    pd.DataFrame({"a": [1, 2], "b": [3, 4]}).to_excel(buffer, index=False)
    buffer.seek(0)
    back = pd.read_excel(buffer)
    assert back.shape == (2, 2)
    print("OK      Excel read/write works")
except Exception as e:
    print(f"FAILED  Excel read/write: {type(e).__name__}: {e}")
```
